```python
model = joblib.load("loan_model.joblib")
```

Se crean nuevos datos para hacer una clasificación

```python
nuevos_datos = pd.DataFrame({
    'income':         [60000],
    'credit_score':   [690],
    'loan_amount':    [20000],
    'years_employed': [10]
})

prediction = model.predict(nuevos_datos)
probability = model.predict_proba(nuevos_datos)

print("Predicción (0 = No aprobado, 1 = Aprobado):", prediction[0])
print(f"Probabilidad de NO aprobación: {probability[0][0]:.2%}")
print(f"Probabilidad de aprobación:    {probability[0][1]:.2%}")
```

```python
# @title Evaluación de Préstamo Interactiva { run: "auto" }

income = 50000 # @param {type:"number"}
credit_score = 388 # @param {"type":"slider","min":300,"max":850,"step":1}
loan_amount = 6000 # @param {type:"number"}
years_employed = 15 # @param {type:"number"}

import pandas as pd

errores = []
if income <= 0: errores.append("Los ingresos deben ser mayores a 0.")
if loan_amount <= 0: errores.append("El monto del préstamo debe ser mayor a 0.")
if years_employed <= 0: errores.append("Los años de empleo no pueden ser negativos o deben ser mayores a 0.")

if errores:
    print("ERROR DE VALIDACIÓN:")
    for error in errores:
        print(f"- {error}")
else:
    nuevos_datos = pd.DataFrame({
        'income': [income],
        'credit_score': [credit_score],
        'loan_amount': [loan_amount],
        'years_employed': [years_employed]
    })

    prediction = model.predict(nuevos_datos)
    probability = model.predict_proba(nuevos_datos)

    print(f"Datos validados para: ${income:,} de ingreso")
    print("-" * 40)
    print(f"Predicción (0 = No aprobado, 1 = Aprobado): {prediction[0]}")
    print(f"Probabilidad de NO aprobación: {probability[0][0]:.2%}")
    print(f"Probabilidad de aprobación:    {probability[0][1]:.2%}")
    print("-" * 40)
```
